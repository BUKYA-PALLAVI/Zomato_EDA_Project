```python
plt.figure(figsize = (16,10))
ax = sns.countplot(data = data , x = data['local address'],palette = 'viridis')
plt.xticks(rotation=90)
plt.show()
```

```python
plt.figure(figsize = (6,6))
ax = sns.countplot(data = data , x = data['online_order'], hue = 'online_order')
plt.legend(title = 'online_order', shadow = True, edgecolor = 'b')
plt.show()
```

```python
plt.figure(figsize = (39,18))
ax = sns.countplot(data = data , x = data['restaurant name'], hue = 'restaurant name')
plt.legend(title = 'restaurant name', shadow = True, edgecolor = 'b')
plt.show()
```

```python
plt.figure(figsize = (36,36))
sns.boxplot(data = data, x = 'area', y = 'local address', palette = 'rainbow')
plt.show()
```
